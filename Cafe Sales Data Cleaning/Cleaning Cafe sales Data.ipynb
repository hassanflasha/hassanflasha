# Cleaning the Price Per unit column
### Since we now know every item it's easy to fill the null values inside "Price Per Unit" column

```python
data["Price Per Unit"].unique()
```

```python
for j in range(len(data)):
    if data.loc[j,"Price Per Unit"] is None or ["nan","ERROR","UNKNOWN"]:
        if data.loc[j,"Item"] == "Coffee":
            data.loc[j,"Price Per Unit"] = '2.0'
        elif data.loc[j,"Item"] == "Tea":
            data.loc[j,"Price Per Unit"] = '1.5'
        elif data.loc[j,"Item"] == "Sandwich":
            data.loc[j,"Price Per Unit"] = '4.0'
        elif data.loc[j,"Item"] == "Salad":
            data.loc[j,"Price Per Unit"] = '5.0'
        elif data.loc[j,"Item"] == "Cake":
            data.loc[j,"Price Per Unit"] = '3.0'
        elif data.loc[j,"Item"] == "Cookie":
            data.loc[j,"Price Per Unit"] = '1.0'
        elif data.loc[j,"Item"] == "Smoothie":
            data.loc[j,"Price Per Unit"] = '4.0'
        elif data.loc[j,"Item"] == "Juice":
            data.loc[j,"Price Per Unit"] = '3.0'
data["Price Per Unit"] = data["Price Per Unit"].astype(float)
data["Price Per Unit"].isna().sum()
```

```python
data.isna().sum()
```

## At first we will need to change every error or unknown values to null values because proceeding to next steps will require the record to be a float type

```python
data.loc[data["Total Spent"].isin(['ERROR', 'UNKNOWN']), "Total Spent"] = np.nan
data["Total Spent"].unique()
data["Total Spent"] = data["Total Spent"].astype(float)
```

```python
data["Quantity"] = pd.to_numeric(data["Quantity"], errors = 'coerce')
data["Quantity"].unique()
```

# Now we can apply the similar method we used on Item column to fill the values inside Quantity and total spent columns

```python
for m in range(3):
    for i in range(len(data)):
        if pd.isna(data.loc[i,"Quantity"]):
            data.loc[i,"Quantity"] = data.loc[i,"Total Spent"] / data.loc[i,"Price Per Unit"]
        elif pd.isna(data.loc[i,"Total Spent"]):
            data.loc[i,"Total Spent"] = data.loc[i,"Quantity"] * data.loc[i,"Price Per Unit"]
        
    
```

# The Remaining 20 Null values inside inside "Quantity" and "Total Spent" is small so we will fill them with statistics Approach

```python
data.describe()
```

```python
# 3 is the mean value for the column
data["Quantity"] = data["Quantity"].fillna(3)
```

```python
data.loc[data["Total Spent"].isna(),"Total Spent"] = data["Quantity"] * data["Price Per Unit"]
```

## Filling the Locaiton and Payment Method Null values with the mode 

```python
data["Location"].unique()
```

```python
data.loc[data["Location"].isin(['UNKNOWN','ERROR']), 'Location'] =np.nan
data["Location"] = data["Location"].fillna(data["Location"].mode().iloc[0])
data.loc[data["Payment Method"].isin(['UNKNOWN','ERROR']),'Payment Method'] = np.nan
data["Payment Method"] = data["Payment Method"].fillna(data["Payment Method"].mode().iloc[0])
```
